# MetroGuard — leakage-safe EDA

This notebook is intentionally thin: reusable data logic lives in `src/metroguard`. Run `metroguard data download` and `metroguard data prepare` first. The February period is described as *no reported failure*, not proven healthy operation.

```python
import json

from metroguard.config import load_config
from metroguard.data import load_features

config = load_config('../configs/default.yaml')
metadata = json.loads(config.data.features_file.with_suffix('.metadata.json').read_text())
features = load_features(config.data.features_file)
metadata
```

```python
display(features[['coverage_ratio']].describe(percentiles=[.01, .1, .5, .9, .99]))
features[['tp2__mean', 'tp3__mean', 'reservoirs__mean', 'motor_current__mean']].plot(
    subplots=True, figsize=(14, 8), title='Causal five-minute signal summaries'
);
```

## Leakage checklist

- No random row split or full-data scaling.
- No centered rolling operation, interpolation, or backfill.
- Windows crossing a gap or an 80% coverage failure are rejected.
- Date and file index are not model inputs.
