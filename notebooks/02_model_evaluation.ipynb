# MetroGuard — locked holdout evaluation

The model designation and fixed alert policy were specified before inspecting March–September. Isolation Forest performs better retrospectively on some metrics, but the TCN is not replaced post hoc.

```python
import json
from pathlib import Path

import pandas as pd

root = Path('..')
report = json.loads((root / 'reports/metrics.json').read_text())
rows = []
for model, values in report['models'].items():
    rows.append({
        'model': model,
        'early_event_recall': values['early_event_recall'],
        'false_alarms_per_day': values['false_alarms_per_day'],
        'failure_interval_pr_auc': values['pr_auc_official_failure_intervals'],
        'time_in_alert_percent': values['time_in_alert_percent'],
    })
pd.DataFrame(rows).set_index('model')
```

```python
events = pd.read_csv(root / 'reports/event_results.csv')
sensitivity = pd.read_csv(root / 'reports/horizon_sensitivity.csv')
cross_event = pd.read_csv(root / 'reports/cross_event_analysis.csv')
display(events[events.model == report['primary_model']])
display(sensitivity[sensitivity.model == report['primary_model']])
display(cross_event[cross_event.model == report['primary_model']])
```

## Interpretation boundary

The event-block analysis is exploratory and unstable: alert thresholds are selected on the other three incident blocks. It is not a second confirmatory holdout. Four events cannot support a narrow confidence claim.
